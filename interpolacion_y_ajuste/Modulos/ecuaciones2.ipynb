```python
import numpy as np
import ecuaciones as edo
```

```python
def f(t,y):
    w = y[0]
    z = y[1]
    f1 = z
    f2 = 3*np.sin(t)+6*z-9*w
    return np.array([f1,f2])
```

```python
t, solucion = edo.Euler(f,0,1,0.25,[0,1])

y_solucion = np.array(solucion)[:,0]
y_prime = np.array(solucion)[:,1]
print(y_solucion)
print(y_prime)
```

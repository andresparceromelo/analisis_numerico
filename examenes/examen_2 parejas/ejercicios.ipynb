$$

$$

$$

$$

$$

# 5. INTERPOLACIÓN DE LAGRANGE

Datos:

| x | y |
|---|---|
| 1 | 2 |
| 2 | 3 |
| 4 | 5 |

Calcular:

$$
P(3)
$$

---

Bases:

$$
L_0=\frac{(x-2)(x-4)}{(1-2)(1-4)}
$$

$$
L_1=\frac{(x-1)(x-4)}{(2-1)(2-4)}
$$

$$
L_2=\frac{(x-1)(x-2)}{(4-1)(4-2)}
$$

Evaluando en:

$$
x=3
$$

$$
L_0=-\frac13
$$

$$
L_1=1
$$

$$
L_2=\frac13
$$

Entonces:

$$
P(3)
=
2\left(-\frac13\right)
+
3(1)
+
5\left(\frac13\right)
$$

$$
P(3)=4
$$

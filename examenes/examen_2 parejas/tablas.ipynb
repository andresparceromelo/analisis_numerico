# FORMULARIO FINAL - ANÁLISIS NUMÉRICO

## Teoría de errores

### Error absoluto

$$
E_a = |x_{real}-x_{aprox}|
$$

### Error relativo

$$
E_r = \frac{|x_{real}-x_{aprox}|}{|x_{real}|}
$$

### Error aproximado porcentual

$$
E_a\%=\left|\frac{x_i-x_{i-1}}{x_i}\right|100
$$

---

# Bisección

Condición:

$$
f(a)f(b)<0
$$

Fórmula:

$$
x_r=\frac{a+b}{2}
$$

Procedimiento:

1. Calcular \(x_r\)
2. Evaluar \(f(x_r)\)
3. Elegir nuevo intervalo
4. Calcular error
5. Repetir

Tabla:

| i | a | b | xr | f(xr) | Ea% |
|---|---|---|---|---|---|

---

# Newton-Raphson

Fórmula:

$$
x_{i+1}=x_i-\frac{f(x_i)}{f'(x_i)}
$$

Tabla:

| i | xi | f(xi) | f'(xi) | xi+1 | Ea% |
|---|---|---|---|---|---|

---

# Jacobi

Sistema:

$$
Ax=b
$$

Despejar cada variable.

Usar únicamente valores de la iteración anterior.

$$
x^{(k+1)}=f(x^{(k)},y^{(k)},z^{(k)})
$$

$$
y^{(k+1)}=f(x^{(k)},y^{(k)},z^{(k)})
$$

$$
z^{(k+1)}=f(x^{(k)},y^{(k)},z^{(k)})
$$

Tabla:

| i | x | y | z | Ea% |
|---|---|---|---|---|

---

# Gauss-Seidel

Misma idea de Jacobi.

Diferencia:

Los valores nuevos se usan inmediatamente.

Tabla:

| i | x | y | z | Ea% |
|---|---|---|---|---|

---

# Interpolación de Lagrange

Polinomio:

$$
P(x)=\sum_{i=0}^{n} y_iL_i(x)
$$

Donde:

$$
L_i(x)=\prod_{j\neq i}\frac{x-x_j}{x_i-x_j}
$$

Caso de tres puntos:

$$
P(x)=y_0L_0+y_1L_1+y_2L_2
$$

---

# Ajuste lineal

Modelo:

$$
y=a+bx
$$

Pendiente:

$$
b=
\frac
{n\sum xy-\sum x\sum y}
{n\sum x^2-(\sum x)^2}
$$

Intercepto:

$$
a=
\frac{\sum y-b\sum x}{n}
$$

# Método de Euler

Problema:

$$
y'=f(x,y)
$$

Fórmula:

$$
y_{i+1}=y_i+h f(x_i,y_i)
$$

Tabla:

| i | xi | yi | f(xi,yi) | yi+1 |
|---|---|---|---|---|

---

# Runge-Kutta Orden 4

$$
k_1=f(x_i,y_i)
$$

$$
k_2=f(x_i+\frac{h}{2},y_i+\frac{h}{2}k_1)
$$

$$
k_3=f(x_i+\frac{h}{2},y_i+\frac{h}{2}k_2)
$$

$$
k_4=f(x_i+h,y_i+hk_3)
$$

Actualización:

$$
y_{i+1}
=
y_i+\frac{h}{6}
(k_1+2k_2+2k_3+k_4)
$$

Tabla:

| i | xi | yi | k1 | k2 | k3 | k4 | yi+1 |
|---|---|---|---|---|---|---|---|

---

# Diferencias rápidas entre Jacobi y Gauss-Seidel

| Jacobi | Gauss-Seidel |
|----------|----------|
| Usa valores viejos | Usa valores nuevos |
| Más lento | Más rápido |
| Fácil paralelización | Difícil paralelización |

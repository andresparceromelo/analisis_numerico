# 📘 Maneras de ingresar una función matemática en Python

En análisis numérico usamos funciones matemáticas constantemente. En Python hay **3 formas** de definirlas, y cada una sirve para cosas distintas.

---

## 1. 🟢 Función clásica `def`

Es la forma más común de definir una función en Python.

````python
def f(x):
    return x**4 - 17*x**2 + 60

# Se evalúa directamente pasando un número
print(f(3))    # → 60 - 153 + 81 = -12
print(f(-7))   # → valor numérico
````

**✅ Ventajas:**
- Evalúa en cualquier número directamente: `f(3.5)`
- Compatible con Bisección, Falsa Posición, Secante
- Ideal para funciones más largas o con lógica interna

**❌ Desventajas:**
- No puedes sacar la derivada simbólica automáticamente

---

## 2. 🟡 Función anónima `lambda`

Es exactamente igual a `def`, pero escrita en una sola línea. Muy útil para funciones cortas.

````python
f = lambda x: x**4 - 17*x**2 + 60

# Se evalúa exactamente igual que def
print(f(3))
print(f(-7))

# También se puede pasar directamente como argumento sin guardarla en variable:
# Biseccion(lambda x: x**4 - 17*x**2 + 60, a, b, tolerancia)
````

**✅ Ventajas:**
- Más compacta que `def`
- Perfecta para pasar como argumento a otra función
- Compatible con Bisección, Falsa Posición, Secante

**❌ Desventajas:**
- No permite derivada simbólica
- No es ideal para funciones complejas o con múltiples líneas

---

## 3. 🔵 Expresión simbólica con SymPy

SymPy trata la función como una **expresión algebraica**, igual que en matemáticas.

````python
import sympy as sp

x = sp.symbols('x')            # declara x como variable simbólica
f = x**4 - 17*x**2 + 60       # f NO es una función, es una expresión

# ✅ Puedes derivar automáticamente:
f_prima = sp.diff(f, x)
print("f'(x) =", f_prima)

# ✅ Puedes evaluar numéricamente con .evalf(subs={...}):
print("f(3) =", f.evalf(subs={x: 3}))

# ❌ Esto NO funciona (error):
# f(3)   → TypeError
````

**✅ Ventajas:**
- Derivada automática con `sp.diff(f, x)`
- Puede simplificar, factorizar, integrar algebraicamente
- Necesaria para **Newton-Raphson** (que requiere la derivada)

**❌ Desventajas:**
- **No se llama como función normal**: `f(3)` da error
- Para evaluarla: `f.evalf(subs={x: 3})`
- No compatible directamente con Bisección o Secante

---

## 🔄 El puente: `sp.lambdify`

Convierte una expresión **SymPy → función Python normal** para poder evaluarla numéricamente.

````python
import sympy as sp
import numpy as np

x = sp.symbols('x')
expr = sp.ln(x**2 + 1) - sp.exp(0.4*x) * sp.cos(sp.pi*x)

# 🔄 Convertir SymPy → función evaluable con numpy
f = sp.lambdify(x, expr, 'numpy')

# Ahora puedes usarla como función normal:
print(f(0.5))       # ✅ número
print(f(-0.434))    # ✅ número
print(f(np.array([0, 0.5, 1, 1.5])))  # ✅ también funciona con arrays
````

---

## 📊 Tabla resumen

| Forma | ¿Evalúa numéricamente? | ¿Derivada automática? | ¿Cuándo usarla? |
|-------|----------------------|----------------------|------------------|
| `def f(x): ...` | ✅ `f(3.5)` | ❌ | Bisección, Secante, Falsa Posición |
| `lambda x: ...` | ✅ `f(3.5)` | ❌ | Igual que `def`, más compacta |
| `sp.symbols` (SymPy) | ⚠️ Solo con `.evalf(subs=...)` | ✅ | Newton-Raphson, álgebra simbólica |
| `sp.lambdify(...)` | ✅ `f(3.5)` | ❌ | Graficar una expr. SymPy, o usarla en otros métodos |

---

## 🎯 Guía rápida para el examen

```
¿Necesitas derivada automática?  →  SymPy
¿Bisección o Secante?            →  lambda o def
¿Graficar una expresión SymPy?   →  lambdify primero, luego graficar
```

````python
# ✅ EJEMPLO COMPLETO: misma función definida de las 3 formas
import sympy as sp

# Forma 1: def
def p_def(x):
    return x**4 - 17*x**2 + 60

# Forma 2: lambda
p_lambda = lambda x: x**4 - 17*x**2 + 60

# Forma 3: SymPy
x_sym = sp.symbols('x')
p_sympy = x_sym**4 - 17*x_sym**2 + 60

# Evaluación
val = 3.0
print(f"def    →  p({val}) = {p_def(val)}")
print(f"lambda →  p({val}) = {p_lambda(val)}")
print(f"sympy  →  p({val}) = {p_sympy.evalf(subs={x_sym: val})}")

# Derivada (solo posible con SymPy)
print(f"\nDerivada simbólica: p'(x) = {sp.diff(p_sympy, x_sym)}")
````
